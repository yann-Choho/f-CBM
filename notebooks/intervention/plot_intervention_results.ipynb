```python
import os
import json 
```

```python
# Directory where run_interventions.ipynb saves its results (relative to the repository root)
RESULTS_DIR = "saved_intervention_results"

paths = {
    "f-CBM (ours)": os.path.join(RESULTS_DIR, "results_intervention_20260305_1.json"),
    "label-free CBM": os.path.join(RESULTS_DIR, "results_intervention_20260306_2.json"),
    # "CT-CBM cos cubed": os.path.join(RESULTS_DIR, "results_intervention_20260309_3.json"),
    "CT-CBM": os.path.join(RESULTS_DIR, "results_intervention_20260310_4.json"),
    "independant-CBM": os.path.join(RESULTS_DIR, "results_intervention_20260311_5.json"),
}

results = {
    name: json.load(open(path))
    for name, path in paths.items()
}
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

for name, data in results.items():
    
    x = data["num_interventions"]
    y = data["accuracy"]
    
    start = y[0]
    
    # rescale so all curves start at 0
    y_rescaled = [(v - start)/start for v in y]
    
    plt.plot(x, y_rescaled, marker="o", label=name)

plt.axhline(0, linestyle="--", linewidth=1)

plt.xlabel("Number of interventions")
plt.ylabel("Accuracy improvement (vs baseline)")
plt.title("Accuracy evolution after interventions (rescaled)")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

for name, data in results.items():

    x = data["num_interventions"]
    y = data["accuracy"]
    
    start = y[0]
    
    # relative normalization
    y_rescaled = [(v - start) / start * 100 for v in y]  # convert to %

    fig.add_trace(
        go.Scatter(
            x=x,
            y=y_rescaled,
            mode="lines+markers",
            name=name,
            marker=dict(size=8),
            line=dict(width=3)
        )
    )

# horizontal baseline
fig.add_hline(
    y=0,
    line_dash="dash",
    line_width=1
)

fig.update_layout(
    xaxis=dict(
        title="<b>Number of Interventions</b>",
        title_font=dict(size=18),
        tickfont=dict(size=13)
    ),
    yaxis=dict(
        title="<b>Accuracy Improvement (%)</b>",
        title_font=dict(size=18),
        tickfont=dict(size=13)
    ),
    template="plotly_white",
    width=1000,
    height=600,
)

fig.show()
```
